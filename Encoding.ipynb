```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (accuracy_score,precision_score,recall_score,f1_score,classification_report,confusion_matrix)

from sklearn.model_selection import train_test_split
```

```python
disease_data = pd.read_csv("newdf.csv")
disease_data
```

```python
disease_data.shape
```

```python
disease_data.isnull().sum()
```

```python
disease_data.duplicated().sum()
```

```python
duplicate_records = disease_data[
    disease_data.duplicated()
]

duplicate_records.head(10)
```

```python
unique_data = disease_data.drop_duplicates().copy()

print("Original Dataset Shape:", disease_data.shape)

print("After Removing Duplicates:", unique_data.shape)
```

```python
unique_data.drop(columns=["132"],inplace=True)
```

```python
feature_columns = unique_data.columns.drop("Disease")
```

```python
unique_data[feature_columns] = unique_data[feature_columns].astype("int8")

unique_data.head()
```

```python
disease_encoder = LabelEncoder()
```

```python
unique_data["Disease_Encoded"] = disease_encoder.fit_transform(unique_data["Disease"])
```

```python
X = unique_data.drop(columns=["Disease", "Disease_Encoded"])
y = unique_data["Disease_Encoded"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=42,stratify=y)
```

```python
random_forest_model = RandomForestClassifier(n_estimators=100,criterion="gini",random_state=42,class_weight="balanced")
```

```python
random_forest_model.fit(X_train, y_train)
```

```python
random_forest_prediction = random_forest_model.predict(X_test)
```

```python
random_forest_accuracy = accuracy_score(y_test,random_forest_prediction)
```

```python
random_forest_precision = precision_score( y_test, random_forest_prediction,average="macro",zero_division=0)
```

```python
print("Accuracy:", random_forest_accuracy)
print("Precision:", random_forest_precision)
```
