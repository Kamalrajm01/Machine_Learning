## Import Necessary Libraries

```python
import pandas as pd

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split 

from sklearn.metrics import mean_absolute_error,mean_squared_error
import statsmodels.formula.api as smf
```

## Import Dataset

```python
Cars_data = pd.read_csv("cars.csv")
Cars_data
```

## Data Preparation

```python
X = Cars_data[["Volume","Weight"]]
y = Cars_data["CO2"]
```

## Model Preparation

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,shuffle=True,random_state=32)
```

## Model Building

```python
Linear_Model = LinearRegression()
```

## Model Training

```python
Linear_Model.fit(X_train,y_train)
```

## Model Testing

```python
y_pred_train = Linear_Model.predict(X_train)
y_pred_train
```

## Model Evaluation

```python
mean_absolute_error(y_train,y_pred_train)
```

```python
mean_squared_error(y_train,y_pred_train)
```

```python
linear_stats_model = smf.ols(formula = 'CO2 ~ Volume + Weight',data=Cars_data)
```

```python
linear_stats_model = linear_stats_model.fit()
```

```python
y_pred = linear_stats_model.predict(X_test)
y_pred
```

```python
print("R-Squared:", linear_stats_model.rsquared)
print("Adjusted R-Squared:", linear_stats_model.rsquared_adj)
```
