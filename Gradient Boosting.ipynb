#### Import Necessary Libraries

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.pipeline import Pipeline

from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
```

#### Data Collection

```python
training_data = pd.read_csv("train.csv")
testing_data = pd.read_csv("test.csv")
```

####  Data Understanding

```python
training_data.isna().sum()
```

#### Data Preprocessing

```python
training_data.drop(columns=["Unnamed: 0"],inplace=True,errors="ignore")

testing_data.drop(columns=["Unnamed: 0"],inplace=True,errors="ignore")
```

```python
training_data["DATE"] = pd.to_datetime(training_data["DATE"],dayfirst=True,errors="coerce")

testing_data["DATE"] = pd.to_datetime(testing_data["DATE"],dayfirst=True,errors="coerce")
```

```python
for dataset in [training_data, testing_data]:

    dataset["YEAR"] = dataset["DATE"].dt.year

    dataset["MONTH"] = dataset["DATE"].dt.month

    dataset["DAY"] = dataset["DATE"].dt.day

    dataset.drop(columns=["DATE"], inplace=True)
```

```python
training_data.drop_duplicates(inplace=True)
```

```python
training_data.dropna(subset=["TAVG"], inplace=True)
```

```python
X = training_data.drop(columns=["TAVG"])

y = training_data["TAVG"]
```

#### Model Building

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=42)
```

```python

gradient_boosting_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),

    ("model", GradientBoostingRegressor(n_estimators=100,learning_rate=0.1,max_depth=3,random_state=42))])
```

#### Model Training

```python
gradient_boosting_model.fit(X_train, y_train)
```

#### Model Evaluation

```python
y_prediction = gradient_boosting_model.predict(X_test)
```

```python
mae = mean_absolute_error(y_test, y_prediction)

mse = mean_squared_error(y_test, y_prediction)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_prediction)
```

```python
print("Gradient Boosting Regression Evaluation")

print("Mean Absolute Error:", mae)

print("Mean Squared Error:", mse)

print("Root Mean Squared Error:", rmse)

print("R2 Score:", r2)
```

#### Model Testing

```python
gradient_boosting_model.fit(X, y)
```

```python
final_prediction = gradient_boosting_model.predict(testing_data)
```
