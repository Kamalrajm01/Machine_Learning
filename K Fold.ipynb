#### Import necessary libraries

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn.model_selection import KFold, cross_val_score
from sklearn.ensemble import RandomForestClassifier
```

#### Data collection

```python
iris_dataset = load_iris()

iris_data = pd.DataFrame(iris_dataset.data,columns=iris_dataset.feature_names)

iris_data['Species'] = iris_dataset.target

iris_data.head()
```

#### Data understanding

```python
iris_data.head()

iris_data.shape

iris_data.columns

iris_data.dtypes

iris_data.isnull().sum()

iris_data.describe()

iris_data['Species'].value_counts()
```

#### Data Preparation

```python
X = iris_data.drop('Species', axis=1)

y = iris_data['Species']
```

#### Creating K-Fold Cross-Validation

```python
kfold = KFold(n_splits=5,shuffle=True,random_state=42)
```

#### Creating Model

```python
random_forest_model = RandomForestClassifier(n_estimators=100,random_state=42)
```

#### Apply K-Fold Cross-Validation

```python
accuracy_scores = cross_val_score(random_forest_model,X,y,cv=kfold, scoring='accuracy')
```

```python
print("Accuracy Scores:", accuracy_scores)
```

```python
print("Average Accuracy:", accuracy_scores.mean())

print("Standard Deviation:", accuracy_scores.std())
```
