#### Import necessary libraries

```python
import pandas as pd
from sklearn import datasets
from sklearn.preprocessing import StandardScaler,MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report,confusion_matrix
```

#### Import dataset

```python
breast_cancer_data = datasets.load_breast_cancer()
```

```python
breast_cancer_data_df = pd.DataFrame(data=breast_cancer_data.data,columns=breast_cancer_data.feature_names)
breast_cancer_data_df
```

```python
breast_cancer_data_df["Target"]= breast_cancer_data.target
breast_cancer_data_df
```

#### Data Understanding

```python
breast_cancer_data_df.shape
```

```python
breast_cancer_data_df.isna().sum()
```

```python
breast_cancer_data_df.dtypes
```

#### Data Preparation

```python
X = breast_cancer_data_df.drop(labels = "Target",axis=1)
y = breast_cancer_data_df["Target"]
```

```python
std_scalar = StandardScaler()

Scaled_X = pd.DataFrame(std_scalar.fit_transform(X = X),columns = breast_cancer_data.feature_names)
Scaled_X
```

```python
Scaled_X.mean(),Scaled_X.std()
```

#### Model Training

```python
X_train,X_test,y_train,y_test=  train_test_split(Scaled_X,y,test_size = 0.20,random_state = 1234,shuffle = True,stratify =y) 
```

```python
dt_model = DecisionTreeClassifier(criterion='gini',max_depth=3, class_weight={0:2,1:1})
```

```python
dt_model.fit(X_train,y_train)
```

#### Model Testing || Model Evaluation

## Training data

```python
y_pred_train = dt_model.predict(X_train)
y_pred_train
```

```python
print(classification_report(y_train,y_pred_train))
print(confusion_matrix(y_train,y_pred_train))
```

## Test Data

```python
y_pred_test= dt_model.predict(X_test)

print(classification_report(y_test,y_pred_test))
print(confusion_matrix(y_test,y_pred_test))
```
