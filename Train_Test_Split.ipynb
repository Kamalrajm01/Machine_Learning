#### Import Necessary libraries

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression 

from sklearn.metrics import accuracy_score
```

#### Import Dataset

```python
Social_Network_Ads_Data = pd.read_csv("Social_Network_Ads.csv")
Social_Network_Ads_Data
```

#### Data Understanding

```python
Social_Network_Ads_Data.shape
```

#### Data Preparation

```python
X = Social_Network_Ads_Data[["Age","EstimatedSalary"]]
y = Social_Network_Ads_Data["Purchased"]
```

#### Model Building

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=32,shuffle=True)
```

#### Model Training

```python
Logistic_model = LogisticRegression()
```

```python
Logistic_model.fit(X_train,y_train)
```

#### Model Testing

#### Training Data

```python
y_pred_train = Logistic_model.predict(X_train)
y_pred_train
```

#### Testing Data

```python
y_pred_test = Logistic_model.predict(X_test)
y_pred_test
```

#### Model Evaluation

```python
accuracy_score(y_train,y_pred_train)
```

```python
accuracy_score(y_test,y_pred_test)
```
